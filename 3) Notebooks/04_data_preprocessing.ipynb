# Stage 06 — Exploratory Data Analysis (EDA)
## Telecom Customer Churn Analysis

This notebook follows the same practical method as the provided snippets, but organizes them into a clean, logical EDA workflow.

**Goal:** understand customer behavior, churn patterns, and relationships that will generate stronger questions for the next diagnostic-analysis phase.

> EDA identifies patterns and associations; it does not prove causation.

## 1. Import Libraries

```python
# Import the main libraries used for analysis and visualization.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Use a consistent visualization style.
sns.set_theme(style="whitegrid")

# Show all columns when inspecting DataFrames.
pd.set_option("display.max_columns", None)
```

**What this cell does:** Imports Pandas/NumPy for analysis and Matplotlib/Seaborn for visualization.

## 2. Load the Cleaned Dataset

```python
# Load the cleaned dataset created in Stage 05.
clean_df = pd.read_csv(r"D:\Data Analytics\Project\2) Dataset\2) Cleaned\telco_churn_cleaned.csv")

print("Dataset loaded successfully.")
print("Shape:", clean_df.shape)
```

**What this cell does:** Loads the Stage 05 cleaned data so EDA does not repeat the cleaning process.

## 3. Dataset Overview

```python
# Basic customer-population statistics.
print("Customers:", len(clean_df))
print("Features:", clean_df.shape[1])
print("Average tenure:", clean_df["tenure"].mean())
print("Average monthly charge:", clean_df["monthly_charges"].mean())
print("Average total charges:", clean_df["total_charges"].mean())
```

**Why:** These values form the baseline against which segment-level results will later be compared.

## 4. Churn Overview

```python
# Churn counts and percentages.
print(clean_df["churn"].value_counts())
print()
print(clean_df["churn"].value_counts(normalize=True).mul(100).round(2))
```

```python
# Visualize the overall churn distribution.
sns.countplot(data=clean_df, x="churn")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

**Why:** Establishes the overall churn baseline and shows the class imbalance in the customer population.

## 5. Churn Rate vs. Churn Count

```python
# Calculate the percentage of customers who churn within each contract type.
contract_churn_table = pd.crosstab(
    clean_df["contract"],
    clean_df["churn"],
    normalize="index"
).mul(100).round(2)

display(contract_churn_table)
```

```python
# Direct churn-rate calculation for each contract type.
contract_churn = (
    clean_df.groupby("contract")["churn_flag"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
    .round(2)
)

display(contract_churn.rename("churn_rate_pct"))
```

**Key idea:** `normalize='index'` calculates the churn percentage within each contract group, which is more meaningful for risk comparison than raw churn counts.

## 6. Tenure EDA

```python
# Statistical summary of tenure.
display(clean_df["tenure"].describe())
```

```python
# Distribution of customer tenure.
sns.histplot(data=clean_df, x="tenure", bins=30)
plt.title("Customer Tenure Distribution")
plt.xlabel("Tenure (Months)")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
# Compare tenure between churned and retained customers.
sns.boxplot(data=clean_df, x="churn", y="tenure")
plt.title("Tenure Distribution by Churn Status")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")
plt.tight_layout()
plt.show()
```

**What to look for:** If churned customers have shorter tenure, early customer lifecycle stages may deserve deeper diagnostic analysis.

## 7. Monthly Charges EDA

```python
# Statistical summary of monthly charges.
display(clean_df["monthly_charges"].describe())
```

```python
# Distribution of monthly charges.
sns.histplot(data=clean_df, x="monthly_charges", bins=30)
plt.title("Monthly Charges Distribution")
plt.xlabel("Monthly Charges")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
# Compare monthly charges by churn status.
sns.boxplot(data=clean_df, x="churn", y="monthly_charges")
plt.title("Monthly Charges by Churn Status")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")
plt.tight_layout()
plt.show()
```

```python
# Compare the complete monthly-charge distributions by churn status.
sns.kdeplot(
    data=clean_df,
    x="monthly_charges",
    hue="churn",
    fill=True
)
plt.title("Monthly Charges Distribution by Churn")
plt.xlabel("Monthly Charges")
plt.ylabel("Density")
plt.tight_layout()
plt.show()
```

## 8. Total Charges

```python
# Compare accumulated total charges between churned and retained customers.
sns.boxplot(data=clean_df, x="churn", y="total_charges")
plt.title("Total Charges by Churn Status")
plt.xlabel("Churn")
plt.ylabel("Total Charges")
plt.tight_layout()
plt.show()
```

**Interpretation caution:** Total charges are strongly related to tenure because charges accumulate over time. Therefore, a difference in total charges should not automatically be interpreted as a churn cause.

## 9. Contract EDA

```python
# Customer counts and percentages by contract.
print(clean_df["contract"].value_counts())
print()
print(clean_df["contract"].value_counts(normalize=True).mul(100).round(2))
```

```python
# Customer distribution by contract.
sns.countplot(data=clean_df, x="contract")
plt.title("Customer Distribution by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# Churn rate by contract.
contract_churn = (
    clean_df.groupby("contract")["churn_flag"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
    .round(2)
)

display(contract_churn.rename("churn_rate_pct"))
```

## 10. Payment Method EDA

```python
# Churn rate by payment method.
payment_churn = (
    clean_df.groupby("payment_method")["churn_flag"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
    .round(2)
)

display(payment_churn.rename("churn_rate_pct"))
```

```python
# Visualize payment-method churn rates.
payment_churn.sort_values().plot(kind="barh", figsize=(9, 5))
plt.title("Churn Rate by Payment Method")
plt.xlabel("Churn Rate (%)")
plt.ylabel("Payment Method")
plt.tight_layout()
plt.show()
```

## 11. Internet Service EDA

```python
# Churn rate by internet service.
internet_churn = (
    clean_df.groupby("internet_service")["churn_flag"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
    .round(2)
)

display(internet_churn.rename("churn_rate_pct"))
```

```python
# Visualize churn rates by internet service.
sns.barplot(x=internet_churn.values, y=internet_churn.index)
plt.title("Churn Rate by Internet Service")
plt.xlabel("Churn Rate (%)")
plt.ylabel("Internet Service")
plt.tight_layout()
plt.show()
```

## 12. Service-Level Analysis

```python
# Optional services to compare.
service_cols = [
    "online_security",
    "online_backup",
    "device_protection",
    "tech_support",
    "streaming_tv",
    "streaming_movies"
]

# Calculate churn rate for each category of each service.
for col in service_cols:
    print("\n", col)
    result = (
        clean_df.groupby(col)["churn_flag"]
        .mean()
        .mul(100)
        .sort_values(ascending=False)
        .round(2)
    )
    print(result)
```

**Why:** This applies the same analysis consistently to every service instead of duplicating six separate blocks of code.

## 13. Demographic EDA

```python
# Churn rate by senior-citizen status.
print("Senior Citizen")
display(
    clean_df.groupby("senior_citizen_label")["churn_flag"]
    .mean().mul(100).round(2)
    .rename("churn_rate_pct")
)

# Churn rate by partner status.
print("Partner")
display(
    clean_df.groupby("partner")["churn_flag"]
    .mean().mul(100).round(2)
    .rename("churn_rate_pct")
)

# Churn rate by dependent status.
print("Dependents")
display(
    clean_df.groupby("dependents")["churn_flag"]
    .mean().mul(100).round(2)
    .rename("churn_rate_pct")
)
```

**Interpretation:** These variables help profile customers, but a difference does not automatically make a variable a strong or actionable churn driver.

## 14. Numeric Comparison by Churn

```python
# Compare mean and median tenure and charges for churned vs. retained customers.
numeric_comparison = (
    clean_df.groupby("churn")[
        ["tenure", "monthly_charges", "total_charges"]
    ]
    .agg(["mean", "median"])
    .round(2)
)

display(numeric_comparison)
```

## 15. Correlation Analysis

```python
# Select the key numeric variables.
numeric_cols = [
    "senior_citizen",
    "tenure",
    "monthly_charges",
    "total_charges",
    "churn_flag"
]

corr = clean_df[numeric_cols].corr()

display(corr.round(2))
```

```python
# Visualize the correlation matrix.
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    center=0
)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

**Important:** Correlation measures association, not causation. In particular, total charges and tenure are naturally related because customers accumulate charges over time.
